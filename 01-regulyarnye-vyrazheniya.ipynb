## 1. Поиск и квантификаторы

Большинство символов в регулярном выражении обозначают сами себя. Особый смысл имеют специальные символы.

**Квантификаторы:**

- `*` — ноль или более повторов
- `+` — один или более повторов
- `?` — ноль или один повтор

````python
import re

print('* :', re.findall(r'ab*', 'a ab abb abbb'))
print('+ :', re.findall(r'ab+', 'a ab abb abbb'))
print('? :', re.findall(r'ab?', 'a ab abb'))
````

## 2. Основные символы

| Символ | Описание |
| --- | --- |
| `.` | любой символ, кроме конца строки |
| `[abcdef]` | любой из перечисленных символов |
| `[A-Za-z]` | интервальный массив символов |
| `[^0-9]` | нецифровые символы |
| `\s` | пробельный символ |
| `\S` | не пробельный символ |
| `\d` | десятичная цифра, то же что `[0-9]` |
| `\D` | нецифра |
| `\w` | алфавитно-цифровой символ или `_` |
| `\W` | всё, что не входит в `\w` |

Альтернатива `(a|b)` — «а либо б».

````python
text = 'Кот cat 42 3.14 _test'
print('\\d :', re.findall(r'\d', text))
print('\\w+ :', re.findall(r'\w+', text))
print('\\S+ :', re.findall(r'\S+', text))
print('[A-Za-z]+ :', re.findall(r'[A-Za-z]+', text))
print('[^0-9]+ :', re.findall(r'[^0-9]+', text))
````

## 3. Практическое задание

Задачи с последовательностью чисел от 1 до 10000. В консоли это делается так:

```bash
seq 1 10000 | grep '^2.*4$'
```

* `^2` — строка начинается с двойки;
* `.*` — между ними любое количество символов;
* `4$` — строка заканчивается на четвёрку.

````python
numbers = [str(n) for n in range(1, 10001)]

task = [s for s in numbers if re.search(r'^2.*4$', s)]
print('начинаются на 2, заканчиваются на 4:', len(task))
print(task[:20])
````

Другие шаблоны из практики на той же последовательности:

````python
print('^(2|3)\\d*(4|5)$ :', [s for s in numbers if re.search(r'^(2|3)\d*(4|5)$', s)][:12])
print('.*[02468]$ (чётные):', [s for s in numbers if re.search(r'.*[02468]$', s)][:12])
print('^..$ (двузначные):', [s for s in numbers if re.search(r'^..$', s)])
````

Поиск дат и телефонов в тексте:

````python
sample = 'Дата 12/05/2026, телефон (7)9123456789, ещё (8)9990001122.'

print('даты:', re.findall(r'\d{2}/\d{2}/\d{4}', sample))
phones = [m.group(0) for m in re.finditer(r'\((7|8)\)\d{10}', sample)]
print('телефоны:', phones)
````

## 4. Группировка и захват

| Конструкция | Описание |
| --- | --- |
| `(xxx)` | найти паттерн и **положить** результат в группу захвата |
| `(?:xxx)` | найти паттерн, но **не** класть в группу |

Доступ к группам: `\1`, `\2`, `\3` или `\g<1>`, `\g<2>`.

````python
pattern = re.compile(r'1(\d)2')
print('замена 1N2 -> sNy:', pattern.sub(r's\1y', '132 142 152 111'))

print(r'группа \1:', re.sub(r'(\w+)@(\w+)', r'\2/\1', 'user@host'))
print(r'именованные \g<n>:', re.sub(r'(?P<a>\w+)@(?P<b>\w+)', r'\g<b>/\g<a>', 'user@host'))
print('фамилия и инициал:', re.sub(r'(.).+\s(.+)', r'\2 \1.', 'Иванов Иван'))
````

## 5. Домашнее задание

Найти в тексте пары «прилагательное на `-ое` + глагол на `-ть/-ти/-чь» в любом порядке:

````python
text = 'красивое шить быстро, синее петь громко, тёмный лес.'
pattern = re.compile(
    r'([а-яё]{3,}[ое]\s[а-яё]+(ть|ти|чь)|'
    r'[а-яё]+(ть|ти|чь)\s[а-яё]{3,}[ое])([^а-яё])'
)
print([m.group(1) for m in pattern.finditer(text)])
````
